### Overview
Multivariate time-series forecasting engine with cyclical temporal encodings, autoregressive lag features, and LightGBM regression for power grid optimization.


## 1. Environment Setup & Data Ingestion
Load hourly grid load, weather telemetry, and renewable generation data.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import joblib
import json

# Load raw time-series
df = pd.read_csv('../data/raw/smart_grid_energy_demand.csv')
df['timestamp'] = pd.to_datetime(df['timestamp'])
print(f'Total Hours Loaded: {len(df):,}')
df.head()
```

## 2. Exploratory Time Series Analysis & Diurnal Profiles
Analyze seasonal and daily diurnal demand patterns across seasons.


```python
plt.figure(figsize=(12, 4))
plt.plot(df['timestamp'][:720], df['grid_total_load_mw'][:720], label='Grid Demand (MW)', color='#2563eb')
plt.plot(df['timestamp'][:720], df['renewable_total_mw'][:720], label='Renewable (MW)', color='#10b981', alpha=0.7)
plt.title('First 30 Days Grid Demand & Renewable Penetration')
plt.xlabel('Date')
plt.ylabel('Power (MW)')
plt.legend()
plt.show()
```
